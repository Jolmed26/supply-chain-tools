```python
%%writefile supply-chain-tools/src/inventory_utils.py

import numpy as numpy

def calculate_safety_stock(max_daily_sales, max_lead_time, avg_daily_sales, avg_lead_time):
  """
  Calculate Safety Stock using the standard formula:
  (Max Daily Sales * Max Lead Time) - (Avg Daily Sales * Avg Lead Time)
  """
  max_usage = max_daily_sales * max_lead_time
  avg_usage = avg_daily_sales * avg_lead_time
  safety_stock = max_usage - avg_usage

  return round(safety_stock, 2)
```
